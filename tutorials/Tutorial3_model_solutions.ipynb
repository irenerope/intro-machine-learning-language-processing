## **Part 1: Rules and matching patterns**


Ordinary characters can be matched by using the character in the pattern. Other characters and meta-references can be used from regular expressions, e.g. `.` is any character, `\b` is boundary of a word, etc.

👋 ⚒ Exercise 1: Write a pattern that matches all occurrences of the word *more* in any capitalization.

The function `findall()` returns all matches to the pattern in the string.

```python
import re

text = "More is not always better, because sometimes it is just more"

pattern_the = r"\b[Mm]ore\b"

print(re.findall(pattern_the, text))
assert re.findall(pattern_the, text) == ["More", "more"], "Expected exactly ['More', 'more']"
print("OK")
```

Let's finish the Exercise 5 from Tutorial 1 on pattern matching. First we need our POS tags and the sentences.

```python
lexicon = {
    "time":      ["NOUN", "VERB"],
    "flies":     ["NOUN", "VERB"],
    "like":      ["VERB", "ADP"],
    "an":        ["DET"],
    "a":         ["DET"],
    "the":       ["DET"],
    "arrow":     ["NOUN"],
    "fruit":     ["NOUN"],
    "banana":    ["NOUN"],
    "i":         ["PRON"],
    "saw":       ["NOUN", "VERB"],
    "her":       ["PRON", "DET"],
    "duck":      ["NOUN", "VERB"],
    "man":       ["NOUN", "VERB"],
    "with":      ["ADP"],
    "telescope": ["NOUN"],
}

sentences = [
    "time flies like an arrow",
    "fruit flies like a banana",
    "i saw her duck",
    "i saw the man with the telescope",
]
```

Next we need to run our solution for creating all possible pos sequences.

```python
def pos_tagging_sequences(sentence):
    sequences = [()]
    words = sentence.lower().split()
    for w in words:
      new_sequences = []
      for seq in sequences:
        for tag in lexicon[w]:
          new_sequences.append(seq +(tag,))
      sequences = new_sequences
    return sequences

for s in sentences:
    tagged_s = pos_tagging_sequences(s)
    print(tagged_s, "sequences:", s)
```

👋 ⚒ Exercise 2: Write rules that can be used to eliminate specific POS sequences, such as


*   The sequence has a VERB
*   A determiner DET is followed by a noun
*   The sequence has no directly adjacent verbs

Use the library re for pattern matching for this purpose. One example is provided. The function `search` finds the first match in a sequence.

```python
import re

def has_verb(tags):
    return re.search(r"VERB", " ".join(tags)) is not None

def det_before_noun(tags):
    return re.search(r"DET VERB", " ".join(tags)) is None

def no_adjacent_verbs(tags):
    return re.search(r"VERB VERB", " ".join(tags)) is None

example_tag_sequence = ('VERB', 'VERB', 'VERB', 'DET', 'NOUN')

print(has_verb(example_tag_sequence))           # True
print(det_before_noun(example_tag_sequence))    # True
print(no_adjacent_verbs(example_tag_sequence))  # False
```

👋 ⚒ Exercise 3: Extract dates like `26.10.2026`. Run the first pattern provided and then make it stricter, that is, day needs to be from 1-31 and month from 1-12.

The function `finditer()`returns an iterator object of match objects for all matches and `group()` returns the entire match date for match groups.

Use a regex tester, e.g. the one from the [W3C school](https://www.w3schools.com/tools/tool_regex.php) to play around with regex patterns.

```python
text_dates = "Meeting on 26.10.2026 and 3.11.2026. Release 1.12.2024 of the software, call 99.99.2026."

def find_dates(pattern, text):
    return [m.group() for m in re.finditer(pattern, text)]

loose = r"\b\d{1,2}\.\d{1,2}\.\d{4}\b"
print("loose: ", find_dates(loose, text_dates))

strict = r"\b(0?[1-9]|[12]\d|3[01])\.(0?[1-9]|1[0-2])\.\d{4}\b"

print("strict:", find_dates(strict, text_dates))
assert find_dates(strict, text_dates) == ["26.10.2026", "3.11.2026", "1.12.2024"]
print("OK")
```

👋 ⚒ Exercise 4: Add a rule that reponds with "Tell me more about your family." when the input contains the words mother or father.

The function `fullmatch()` checks whether the entire string matches the pattern.

```python
rules = [
    (r"I am (.*)",   r"Why are you \1?"),
    (r"I feel (.*)", r"Why do you feel \1?")
]

def respond(utterance):
    for pattern, answer in rules:
        m = re.fullmatch(pattern, utterance, flags=re.IGNORECASE)
        if m:
            return m.expand(answer)
    return "Please go on."

for u in ["I am sad", "I feel tired today", "My mother called", "I met my father yesterday", "Hello"]:
    print(u, "->", respond(u))

# Your code here
```
